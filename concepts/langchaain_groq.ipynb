# Chaining

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


output = StrOutputParser()

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that translates {language}. Translate the user sentence."),
    ("human", "{Query}")
])
```

```python
def transform_case(res:str) -> str:
    return res.upper()
```

```python
chains = prompt | llm | output | transform_case
chains
```

```python
res = chains.invoke({
    "language": "Hinglish",
    "Query": "I love programming."})

res
```

# Memory

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    groq_api_key=os.getenv("GROQ_API_KEY")
)
```

```python
prompts = [
    {
        "role":"user",
    "content": "What is my name?"
    }
]

res = llm.invoke(prompts)
```

```python
res.content
```

```python
history = []

while True:
    query = input("User: ")

    if query.lower() in ["exit", "bye", "quit"]:
        print("GoodBye!")
        break

    history.append({"role": "user", "content": query})
    print("User:", query)

    res = llm.invoke(history)
    history.append({"role": "ai", "content": res.content})
    print("AI:", res.content, "\n")
```

# LLM Respons in real time

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    groq_api_key=os.getenv("GROQ_API_KEY"),
    streaming=True
)
```

```python
question = "Can you explain me Gen-AI?"
```

```python
response = llm.stream(question)

for chunk in response:
    print(chunk.content, end="")
```
