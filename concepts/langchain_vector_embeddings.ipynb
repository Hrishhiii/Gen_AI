```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "I love playing football with my friends on weekends.",
    "Football is my favorite sport, and I usually play it every Sunday.",
    "I enjoy watching movies, especially science-fiction and thriller films.",
    "Machine learning allows computers to learn patterns from data.",
    "Python is widely used for building machine learning applications."
]
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vector = embeddings.embed_documents(documents)
```

```python
print(len(vector[0]))
```

# Store vector embeddings in Vector DB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(texts=documents, embedding=embeddings, persist_directory="./chroma_db")
```

```python
query = "What is the best sport to play on weekends?"
result = vector_store.similarity_search(query,k=2)
result
```
